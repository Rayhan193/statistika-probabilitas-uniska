```python
# STEP 1: Load Dataset
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

URL = "https://raw.githubusercontent.com/Rayhan193/statistika-probabilitas-uniska/refs/heads/main/steam.csv"
df_my = pd.read_csv(URL)
print(f"Shape: {df_my.shape}")
df_my.head()
```

```python
# STEP 2: Missing & Duplicate Check
print(df_my.isnull().sum())
print(f"Duplicates: {df_my.duplicated().sum()}")
```

```python
# STEP 3: Central Tendency
numeric = df_my.select_dtypes(include='number').columns.tolist()
print(df_my[numeric].describe())
```

```python
# STEP 4: Dispersion
print({c: {'std': df_my[c].std(), 'IQR': df_my[c].quantile(0.75) - df_my[c].quantile(0.25),
           'CV': df_my[c].std() / df_my[c].mean() * 100} for c in numeric})
```

```python
#STEP 5: Univariate Viz — min 3 histogram/boxplot
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Mengatur tema visualisasi
sns.set_theme(style="whitegrid")

# Menentukan kolom analisis
kolom_analisis = ['price', 'positive_ratings', 'negative_ratings']
colors = ['#4D96FF', '#FF6B6B', '#6BCB77']

# 2. Membuat Grid Boxplot (3 Baris, 1 Kolom atau sebaliknya)
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
fig.suptitle('Deteksi Outlier (Boxplot - Skala Log untuk Ratings)', fontweight='bold', fontsize=16, y=1.05)

for ax, col, color in zip(axes, kolom_analisis, colors):
    if 'ratings' in col:
        # Boxplot dengan skala log untuk visualisasi outlier yang lebih rapi
        sns.boxplot(y=np.log1p(df_my[col]), ax=ax, color=color, width=0.4)
        ax.set_title(f'Boxplot Log1p {col.replace("_", " ").title()}', fontweight='bold')
        ax.set_ylabel(f'Log1p({col.title()})')
    else:
        sns.boxplot(y=df_my[col], ax=ax, color=color, width=0.4)
        ax.set_title(f'Boxplot {col.title()}', fontweight='bold')
        ax.set_ylabel(col.title())

plt.tight_layout()
plt.show()
```

```python
# 2x2 grid: histogram untuk 3 numerik + countplot platforms
import matplotlib.pyplot as plt

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Kolom numerik pilihan yang menarik untuk dianalisis
pilihan_numerik = ['price', 'positive_ratings', 'negative_ratings']
colors = ['#6366F1', '#EC4899', '#10B981']

for ax, col, color in zip(axes.flatten()[:3], pilihan_numerik, colors):
    # Menggunakan log scale karena nilai ratings sangat bervariasi
    if 'ratings' in col:
        import numpy as np
        # Menghindari log(0) dengan log1p
        np.log1p(df_my[col]).hist(bins=25, ax=ax, color=color, edgecolor='black', alpha=0.8)
        ax.set_title(f'Distribusi Log1p {col}', fontweight='bold')
    else:
        df_my[col].hist(bins=25, ax=ax, color=color, edgecolor='black', alpha=0.8)
        ax.axvline(df_my[col].mean(), color='red', linestyle='--', linewidth=2, label=f'Mean={df_my[col].mean():.1f}')
        ax.axvline(df_my[col].median(), color='orange', linestyle='--', linewidth=2, label=f'Median={df_my[col].median():.1f}')
        ax.set_title(f'Distribusi {col}', fontweight='bold')
        ax.legend()

    ax.set_xlabel(col)
    ax.set_ylabel('Frequency')

# Distribusi Platform (sebagai pengganti Gender)
ax = axes[1, 1]
# Kita ambil 3 kategori platform teratas saja agar visualisasinya rapi
counts = df_my['platforms'].value_counts().head(3)
bars = ax.bar(counts.index, counts.values, color=['#F59E0B', '#6366F1', '#10B981'], edgecolor='black', alpha=0.85)
ax.set_title('Top 3 Platform Distribution', fontweight='bold')
ax.set_ylabel('Count')
for bar, val in zip(bars, counts.values):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 100,
            f'{val} ({val/len(df_my)*100:.1f}%)', ha='center', fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Mengatur tema visualisasi
sns.set_theme(style="whitegrid")

# 1. Scatter Plot: Price vs Positive Ratings (dengan skala logaritmik)
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df_my,
    x='price',
    y='positive_ratings',
    alpha=0.6,
    color='#FF6B6B',
    edgecolor='black',
    linewidth=0.5
)
plt.yscale('log') # Skala logaritmik karena perbedaan nilai rating sangat ekstrim
plt.title('Scatter Plot: Price vs Positive Ratings (Log Scale)', fontweight='bold', fontsize=14, pad=15)
plt.xlabel('Price (USD)', fontsize=12, fontweight='bold')
plt.ylabel('Positive Ratings (Log Scale)', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# 2. Heatmap Korelasi untuk variabel numerik terpilih
plt.figure(figsize=(10, 8))
kolom_numerik_pilihan = ['price', 'positive_ratings', 'negative_ratings', 'achievements', 'average_playtime', 'median_playtime']
korelasi = df_my[kolom_numerik_pilihan].corr()

sns.heatmap(
    korelasi,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5,
    square=True,
    cbar_kws={"shrink": .8}
)
plt.title('Heatmap Korelasi Variabel Numerik Utama', fontweight='bold', fontsize=14, pad=15)
plt.tight_layout()
plt.show()
```

### STEP 7: 3-5 Insight Tertulis

Berdasarkan hasil analisis deskriptif, analisis sebaran (univariate), dan korelasi (bivariate) pada dataset Steam, berikut adalah beberapa insight utama yang berhasil ditemukan:

1. **Dominasi Pasar Game Murah dan Gratis**  
   Mayoritas game yang didistribusikan di platform Steam berada pada rentang harga yang sangat terjangkau. Hal ini ditunjukkan oleh nilai median harga game yang hanya sebesar 3.99USD,jauh di bawah nilai rata−ratanya (6.08 USD). Distribusi harga ini sangat condong ke kanan (skewed right), mengindikasikan bahwa game premium berharga tinggi merupakan minoritas di platform ini.

2. **Hubungan Erat Antara Popularitas dan Masukan Pengguna**  
   Terdapat korelasi positif yang kuat (**0.76**) antara jumlah penilaian positif (`positive_ratings`) dan penilaian negatif (`negative_ratings`). Ini menunjukkan bahwa game yang menarik perhatian banyak pemain secara alami akan mengumpulkan kedua jenis rating dalam jumlah besar. Popularitas (*exposure*) yang tinggi memicu volume interaksi yang besar dari kedua belah pihak (pemain yang puas maupun yang kecewa).

3. **Harga Tidak Menjamin Popularitas Ataupun Waktu Bermain**  
   Koefisien korelasi antara harga (`price`) dengan penilaian positif (`0.04`) serta rata-rata waktu bermain (`0.05`) sangat mendekati nol. Hal ini mengindikasikan bahwa game yang lebih mahal tidak serta-merta mendapatkan respons yang lebih positif atau dimainkan lebih lama oleh pengguna. Banyak game gratis (*free-to-play*) atau murah justru memiliki tingkat keterlibatan (*engagement*) dan rating yang jauh lebih tinggi dibandingkan game premium.

4. **Loyalitas Ekstrim pada Sebagian Kecil Game**  
   Analisis dispersi menunjukkan nilai standar deviasi yang sangat tinggi pada variabel waktu bermain (`average_playtime` dan `median_playtime`). Sebagian besar game memiliki waktu bermain median sebesar 0 jam (belum pernah dimainkan secara signifikan oleh mayoritas pembeli), namun nilai maksimumnya mencapai **190.625 jam**. Ini menunjukkan adanya fenomena *long-tail*, di mana hanya ada segelintir game populer yang berhasil mempertahankan loyalitas pemainnya dalam jangka waktu yang sangat ekstrem.
